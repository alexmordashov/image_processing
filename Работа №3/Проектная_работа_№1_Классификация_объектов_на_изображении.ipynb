```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_names = ['coins1.jpg', 'coins2.jpg', "coins3.jpg", "coins4.jpg"]

fig, axes = plt.subplots(nrows=len(image_names), ncols=4, figsize=(15, 15))
for i, image_name in enumerate(image_names):
    img = cv2.imread(image_name)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    gray_blurred = cv2.GaussianBlur(gray, (5, 5), 0)
    c_anny = cv2.Canny(gray_blurred, 50, 150, apertureSize=3)

    axes[i,0].imshow(gray, cmap='gray')
    axes[i,0].set_title(f'Оттенки серого: {image_name}')
    axes[i,1].imshow(gray_blurred, cmap='gray')
    axes[i,1].set_title(f'Размытие Гаусса: {image_name}')
    axes[i,2].imshow(c_anny, cmap='gray')
    axes[i,2].set_title(f'Края Кэнни: {image_name}')
    axes[i,3].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    axes[i,3].set_title(f'Изначальный вариант: {image_name}')
plt.show()
```

```python
DETECTION_PARAMS = {
    'coins1.jpg': dict(preprocess='gauss', dp=1.2, minDist=200, param1=50, param2=150, minRadius=150, maxRadius=250),
    'coins2.jpg': dict(preprocess='gauss', dp=1.2, minDist=200, param1=50, param2=150, minRadius=100, maxRadius=250),
    'coins3.jpg': dict(preprocess='gauss', dp=1.2, minDist=200, param1=50, param2=150, minRadius=100, maxRadius=250),
    'coins4.jpg': dict(preprocess='gauss', dp=1.2, minDist=200, param1=50, param2=150, minRadius=100, maxRadius=250),
}

def detect_circles(image, params):
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    gray_blurred = cv2.GaussianBlur(gray, (3, 3), 0)
    work = gray_blurred
    circles = cv2.HoughCircles(work,
                               cv2.HOUGH_GRADIENT,
                               params['dp'],
                               params['minDist'],
                               param1=params['param1'],
                               param2=params['param2'],
                               minRadius=params['minRadius'],
                               maxRadius=params['maxRadius'])
    return np.round(circles[0, :]).astype('int')

def detect_lines(image, cx, cy, r, canny_low=50, canny_high=150,
                           hough_thresh=20, min_line_len=10, max_line_gap=5):
    h, w = image.shape[:2]
    mask = np.zeros((h, w), dtype=np.uint8)
    cv2.circle(mask, (cx, cy), r, 255, -1)
    x1 = max(cx - r, 0)
    y1 = max(cy - r, 0)
    x2 = min(cx + r, w)
    y2 = min(cy + r, h)
    roi = image[y1:y2, x1:x2]
    roi_mask = mask[y1:y2, x1:x2]

    roi_gray = cv2.cvtColor(roi, cv2.COLOR_BGR2GRAY) if roi.ndim == 3 else roi
    roi_blur = cv2.GaussianBlur(roi_gray, (3, 3), 0)
    roi_edges = cv2.Canny(roi_blur, canny_low, canny_high)
    roi_edges = cv2.bitwise_and(roi_edges, roi_edges, mask=roi_mask)
    lines = cv2.HoughLinesP(
        roi_edges,
        rho=1,
        theta=np.pi / 180,
        threshold=hough_thresh,
        minLineLength=min_line_len,
        maxLineGap=max_line_gap
    )
    result_lines = []
    if lines is not None:
        lines = lines.reshape(-1, 4)
        for lx1, ly1, lx2, ly2 in lines:
            result_lines.append([int(lx1) + x1, int(ly1) + y1,
                                 int(lx2) + x1, int(ly2) + y1])
    full_edges = np.zeros((h, w), dtype=np.uint8)
    full_edges[y1:y2, x1:x2] = roi_edges

    return result_lines, full_edges

def count_lines(lines):
    vertical_count = 0
    horizontal_count = 0
    for (x1, y1, x2, y2) in lines:
        dx, dy = x2 - x1, y2 - y1
        angle = np.degrees(np.arctan2(dy, dx)) % 180
        if angle < 20 or angle > 160:
            horizontal_count += 1
        elif 70 < angle < 110:
            vertical_count += 1
    return vertical_count, horizontal_count

def _vote(value, refs, tol):
    in_range = [(nom, abs(value - ref)) for nom, ref in refs if abs(value - ref) <= tol]
    if in_range:
        return min(in_range, key=lambda x: (x[1], x[0]))[0]
    return min(refs, key=lambda kv: abs(value - kv[1]))[0]

COINS = {
    'coins1.jpg': [(1, 164, 169, 98), (2, 186, 217, 162), (5, 199, 190, 168), (10, 176, 589, 93)],
    'coins2.jpg': [(0.1, 117, 54, 100), (0.5, 140, 109, 126), (1, 148, 171, 119),
                  (10, 158, 463, 53), (5, 174, 133, 185), (2, 153, 128, 124)],
    'coins3.jpg': [(1, 107, 88, 39), (2, 122, 92, 91), (5, 136, 104, 141)],
    'coins4.jpg': [(1, 158, 125, 83), (2, 179, 178, 109), (5, 195, 120, 138)],
}
TOLERANCE = {"radius": 8, "vertical": 40, "horizontal": 40}

def classify_coin(image_name, radius, vertical_count, horizontal_count):
    table = COINS[image_name]
    votes_by_criterion = {
        "radius":     _vote(radius, [(n, r) for n, r, v, h in table], TOLERANCE["radius"]),
        "vertical":   _vote(vertical_count, [(n, v) for n, r, v, h in table], TOLERANCE["vertical"]),
        "horizontal": _vote(horizontal_count, [(n, h) for n, r, v, h in table], TOLERANCE["horizontal"]),
    }

    vote_counts = {nom: 0 for nom, *_ in table}
    for nom in votes_by_criterion.values():
        vote_counts[nom] += 1

    if max(vote_counts.values()) == 1:
        return votes_by_criterion["radius"], votes_by_criterion

    max_count = max(vote_counts.values())
    tied = [nom for nom, cnt in vote_counts.items() if cnt == max_count]

    if len(tied) == 1:
        winner = tied[0]
    else:
        winner = (votes_by_criterion["radius"]
                  if votes_by_criterion["radius"] in tied
                  else min(tied))

    return winner, votes_by_criterion

grand_total = 0
grand_count = 0

for image_name in image_names:
    img = cv2.imread(image_name)
    params = DETECTION_PARAMS[image_name]
    circles = detect_circles(img, params)
    result = img.copy()
    all_edges = np.zeros(img.shape[:2], dtype=np.uint8)
    coin_values = []
    coin_info = []

    for i, (cx, cy, r) in enumerate(circles):
        lines, edges = detect_lines(img, cx, cy, r)
        vertical_count, horizontal_count = count_lines(lines)
        nominal, votes = classify_coin(image_name, r, vertical_count, horizontal_count)
        coin_values.append(nominal)
        coin_info.append((i, cx, cy, r, vertical_count, horizontal_count, nominal))
        cv2.circle(result, (cx, cy), r, (0, 255, 0), 2)
        for (x1, y1, x2, y2) in lines:
            cv2.line(result, (x1, y1), (x2, y2), (0, 0, 255), 1)
        all_edges = cv2.bitwise_or(all_edges, edges)

        print(f"Монета {i}: r={r}, V={vertical_count}, H={horizontal_count}, {nominal} руб.")

    total = sum(coin_values)
    for (i, cx, cy, r, v, h, nominal) in coin_info:
        cv2.putText(result, str(nominal), (cx - 30, cy - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 1.5, (255, 0, 0), 3)
    cv2.putText(result, "Total: " + str(total), (10, 40),
                cv2.FONT_HERSHEY_SIMPLEX, 1.2, (255, 0, 0), 3)
    print("Найдено монет:", len(coin_values))
    print("Номиналы монет:", coin_values)
    print("Cумма:", total, "рублей")
    print()

    grand_total += total
    grand_count += len(coin_values)
    plt.figure(figsize=(20, 10))
    plt.subplot(1, 3, 1)
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title('Изначальный вариант')
    plt.subplot(1, 3, 2)
    plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
    plt.title(f'Найдено монет: {len(coin_values)}. Сумма: {total} руб.')
    plt.title('Края обнаруженных монет')
    plt.show()
```

```python
rho_resolution = 1            # Шаг изменения rho в пикселях
theta_resolution = np.pi / 180  # Шаг изменения theta в радианах
threshold = 170              # Минимальное число пересечений, необходимых для обнаружения линии

# Обнаружение прямых
lines = cv2.HoughLines(canny_img, rho_resolution, theta_resolution, threshold)

img_lines = img.copy()

# Проверка, найдены ли линии
if lines is not None:
    for line in lines:
        rho, theta = line[0]
        # Преобразование параметров обратно в координаты точек
        a = np.cos(theta)
        b = np.sin(theta)
        x0 = a * rho
        y0 = b * rho
        # Определение координат точки на прямой
        x1 = int(x0 + 1000 * (-b))
        y1 = int(y0 + 1000 * a)
        x2 = int(x0 - 1000 * (-b))
        y2 = int(y0 - 1000 * a)
        # Рисование линии на изображении
        cv2.line(img_lines, (x1, y1), (x2, y2), (0, 0, 255), 2)
else:
    print("Прямые не найдены")

# Отображение результатов
plt.figure(figsize=(15, 10))
plt.subplot(1, 2, 1)
plt.title('Границы (Метод Кэнни)')
plt.imshow(img)
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title('Обнаруженные прямые (Преобразование Хафа)')
plt.imshow(cv2.cvtColor(img_lines, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```
